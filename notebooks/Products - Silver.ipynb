```python
import pyspark.sql.functions as F
```

##### Read From Silver

```python
df = spark.read.table('supply_chain_project.bronze.products')
```

##### Perform Required Pre-Processing

```python
df = df.withColumn("unit_margin",F.col("unit_price")-F.col("unit_cost"))
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("supply_chain_project.silver.products")
```
