##### Load the Customers Data

```python
df = spark.read.format('csv').option('header','true').option("inferschema", 'true').load('/Volumes/supply_chain_project/landing_data/landing_data_volume/Customers.csv')
```

##### Check for Nulls

```python
display(df.count())
```

```python
display(df.describe())
```

##### Write the data to bronze layer

```python
%sql
drop table if exists supply_chain_project.bronze.customers
```

```python
df.write.format("delta").mode("append").saveAsTable("supply_chain_project.bronze.customers")
```
