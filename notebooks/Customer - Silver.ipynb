```python
import pyspark.sql.functions as F
```

##### Read Data From Bronze

```python
df = spark.read.table("supply_chain_project.bronze.customers")
```

##### Required Transformations

```python
df = df.withColumn("first_name", F.split("name"," ")[0])
df = df.withColumn("last_name", F.split("name"," ")[1])
df = df.withColumn("email",F.lower("email"))
df = df.withColumn("is_organic_search", F.when(F.col("signup_channel")=='organic', F.lit(True)).otherwise(F.lit(False)))
```

```python
df.display()
```

##### Write Data

```python
df.write.format("delta").mode("overwrite").option("mergeSchema","true").saveAsTable("supply_chain_project.silver.customers")
```
