##### Read Data

```python
df = spark.read.format("csv").option("header","true").option("inferschema","true").load("/Volumes/Supply_Chain_Project/Landing_Data/Landing_Data_Volume/Orders.csv")
```

```python
import pyspark.sql.functions as F
df.agg(F.countDistinct("order_id").alias('unique_orders')).display()
```

##### Write Data

```python
df.write.format("delta").mode("append").saveAsTable("supply_chain_project.bronze.orders")
```

```python
spark.read.table("supply_chain_project.bronze.orders").summary().display()
```
