##### Prepare Gold Data

- Dim Customers
- Dim Products
- Fact Orders

Read Customer from Delta Lake


```python
customer = spark.read.format("delta").table("supply_chain_project.silver.customers")
# display(customer)

#Here are the DataFrames in the session:
#
```

```python
customer.write.format("delta").mode("overwrite").option("mergeSchema","true").saveAsTable("supply_chain_project.gold.fact_customers")
```

Read Products File

```python
products = spark.read.table("supply_chain_project.silver.products")
```

```python
products.write.format("delta").mode("overwrite").option("mergeSchema","true").saveAsTable("supply_chain_project.gold.fact_products")
```

##### Fact Orders

```python
spark.read.table("supply_chain_project.silver.orders").write.format("delta").mode("overwrite").option("mergeSchema","true").saveAsTable("supply_chain_project.gold.fact_orders")
```
