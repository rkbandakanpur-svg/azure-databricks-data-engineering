##### Read Data

```python
df = spark.read.format("csv").option("header", "true").option("inferschema","true").load("/Volumes/Supply_Chain_Project/Landing_Data/Landing_Data_Volume/Products.csv")
```

##### Write Data

```python
df.write.format("delta").mode("append").saveAsTable("Supply_Chain_Project.Bronze.Products")
```
