```python
import pyspark.sql.functions as F
```

##### Read Bronze File

```python
df = spark.read.table("supply_chain_project.bronze.orders")
```

```python
df = df.withColumn("total_price", F.col("quantity")*F.col("unit_price"))
df = df.withColumn("status", F.lower(F.col("status")))
df = df.fillna("NA")
df = df.withColumn("Key",F.concat(F.col("order_id"),F.col("product_id")))  
```

```python
df.describe().display()
```

##### Save File

```python
df.write.format("delta").mode("overwrite").option("mergeSchema", 'true').saveAsTable("supply_chain_project.silver.orders")
```
